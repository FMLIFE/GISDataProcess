```python
import os
from openpyxl import load_workbook

# 设置你的文件夹路径
folder_path = "E:\成果库\excel表0822"  # 请将此路径替换为你的实际文件夹路径

# 定义要添加的前缀
prefix = "b_"

def add_prefix_to_sheets(path):
    """
    遍历指定文件夹下的所有 .xlsx 文件，并为每个文件的所有工作表名称添加前缀。
    """
    try:
        # 确保文件夹存在
        if not os.path.exists(path):
            print(f"错误: 文件夹路径 '{path}' 不存在。")
            return
        numexel = 0
        # 遍历文件夹中的所有文件
        for filename in os.listdir(path):
            
            # 检查文件是否为 .xlsx 格式
            if filename.endswith(".xlsx"):
                file_path = os.path.join(path, filename)
                print(f"正在处理文件: {filename}")

                try:
                    # 加载工作簿
                    workbook = load_workbook(file_path)

                    # 遍历工作簿中的所有工作表
                    for sheet in workbook.worksheets:
                        original_name = sheet.title
                        new_name = prefix + original_name

                        # 检查新名称是否有效，避免名称过长或重复
                        if len(new_name) <= 31 and new_name not in [s.title for s in workbook.worksheets if s != sheet]:
                            sheet.title = new_name
                            print(f"  - 已将工作表 '{original_name}' 重命名为 '{new_name}'")
                        else:
                            print(f"  - 无法重命名工作表 '{original_name}'。新名称 '{new_name}' 无效或已存在。")

                    # 保存工作簿
                    workbook.save(file_path)
                    print(f"已成功保存更改到 {filename}\n")
                    numexel = numexel + 1

                except Exception as e:
                    print(f"  - 处理文件 {filename} 失败。错误信息: {e}\n")
        print(f"已处理完所有excel表格，数量{numexel}\n")
    except Exception as e:
        print(f"发生错误: {e}")

# 调用函数
if __name__ == "__main__":
    add_prefix_to_sheets(folder_path)
```
