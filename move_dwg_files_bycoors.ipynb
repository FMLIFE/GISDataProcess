```python
import os
import shutil

def move_files_by_list(file_list_path, source_directory, destination_directory):
    """
    根据txt文件中列出的文件名，将对应的文件从源目录移动到目标目录。

    Args:
        file_list_path (str): 包含文件名的txt文件路径，每行一个文件名。
        source_directory (str): 文件的原始目录路径。
        destination_directory (str): 文件将被移动到的目标目录路径。
    """

    moved_files = []
    not_found_files = []
    failed_moves = []

    print(f"正在读取文件列表：{file_list_path}")
    print(f"源目录：{source_directory}")
    print(f"目标目录：{destination_directory}\n")

    # 确保目标目录存在，如果不存在则创建
    os.makedirs(destination_directory, exist_ok=True)

    try:
        with open(file_list_path, 'r', encoding='utf-8') as f:
            for line_num, filename in enumerate(f, 1):
                filename = filename.strip() # 移除行首尾的空白符（包括换行符）
                if not filename: # 跳过空行
                    continue

                source_path = os.path.join(source_directory, filename)
                destination_path = os.path.join(destination_directory, filename)

                if os.path.exists(source_path):
                    try:
                        print(f"[{line_num}] 正在移动：'{filename}' 到 '{destination_directory}'...")
                        shutil.move(source_path, destination_path)
                        moved_files.append(filename)
                        print(f"    成功移动：'{filename}'")
                    except Exception as e:
                        failed_moves.append(f"{filename} (错误: {e})")
                        print(f"    移动失败：'{filename}' - {e}")
                else:
                    not_found_files.append(filename)
                    print(f"[{line_num}] 未找到文件：'{filename}'")

    except FileNotFoundError:
        print(f"错误：文件列表文件 '{file_list_path}' 未找到。请检查路径。")
        return
    except Exception as e:
        print(f"读取文件列表时发生错误：{e}")
        return

    print("\n--- 移动结果总结 ---")
    if moved_files:
        print(f"成功移动了 {len(moved_files)} 个文件：")
        for f in moved_files:
            print(f"  - {f}")
    else:
        print("没有文件被成功移动。")

    if not_found_files:
        print(f"\n以下 {len(not_found_files)} 个文件在源目录中未找到：")
        for f in not_found_files:
            print(f"  - {f}")

    if failed_moves:
        print(f"\n以下 {len(failed_moves)} 个文件移动失败：")
        for f in failed_moves:
            print(f"  - {f}")

    print("\n操作完成。")

# --- 使用示例 ---
if __name__ == "__main__":
    # 1. 准备你的文件列表TXT文件
    #    例如：创建一个名为 'file_names.txt' 的文件，内容如下：
    #    file1.txt
    #    image.png
    #    document.pdf
    #    another_file.zip

    # 2. 准备你的源文件夹
    #    例如：在你的脚本同级目录下创建一个 'source_files' 文件夹，
    #    并将 'file1.txt', 'image.png' 等文件放入其中。

    # 3. 指定文件列表TXT文件的路径
    file_list_file = r'D:\CAD处理\矢量恢复_dwg\Beijing_1954_3_Degree_GK_CM_105E.txt'

    # 4. 指定源目录路径（文件当前所在的目录）
    #    你可以使用 os.getcwd() 获取当前脚本所在目录
    #    或者指定一个绝对路径，例如: r'C:\Users\YourUser\Documents\MyFiles'
    source_folder = os.path.join(r'D:\CAD处理\矢量恢复_dwg\结果')

    # 5. 指定目标目录路径（文件将被移动到的新目录）
    #    如果该目录不存在，脚本会自动创建
    destination_folder = os.path.join(r'D:\CAD处理\矢量恢复_dwg\GDB')

    # 调用函数执行文件移动操作
    move_files_by_list(file_list_file, source_folder, destination_folder)
```
