```python
import pandas as pd
import io

# Load the two CSV files into pandas DataFrames
try:
    df_a = pd.read_csv(r"E:\csv-utf8\YLJGA.csv")
    df_p = pd.read_csv(r"E:\csv-utf8\YLJGP.csv")
except Exception as e:
    print(f"Error loading files: {e}")
    # Handle the error or exit
    exit()

# 1. Statistical analysis of both files
# Combine the two dataframes
df_combined = pd.concat([df_a, df_p])
# Group by XZQ and CCN, and count the total number of records
grouped_combined = df_combined.groupby(['XZQ', 'CCN']).size().reset_index(name='Total_Count')
grouped_combined.to_csv('E:\csv-utf8\结果\combined_record_count.csv', index=False, encoding='utf-8-sig')

# 2. Statistical analysis of YLJGA file
# Group by XZQ and CCN, and count the total number of records
grouped_a_count = df_a.groupby(['XZQ', 'CCN']).size().reset_index(name='Total_Count')
grouped_a_count.to_csv('E:\csv-utf8\结果\yljga_record_count.csv', index=False, encoding='utf-8-sig')

# 3. Statistical analysis of YLJGA file, grouping by XZQ and CCN
# Calculate the sum of '椭球面积' and the total number of records
grouped_a_metrics = df_a.groupby(['XZQ', 'CCN']).agg(
    椭球面积_sum=('椭球面积', 'sum'),
    Record_Count=('OBJECTID', 'count')
).reset_index()
grouped_a_metrics.to_csv('E:\csv-utf8\结果\yljga_metrics.csv', index=False, encoding='utf-8-sig')

# 4. Statistical analysis of YLJGA file, grouping by XZQ, CCN, and TYPE
# Calculate the sum of '椭球面积'
grouped_a_type_metrics = df_a.groupby(['XZQ', 'CCN', 'TYPE']).agg(
    椭球面积_sum=('椭球面积', 'sum')
).reset_index()
grouped_a_type_metrics.to_csv('E:\csv-utf8\结果\yljga_type_metrics.csv', index=False, encoding='utf-8-sig')

print("All tasks completed. The following CSV files have been created:")
print("- combined_record_count.csv (Task 1)")
print("- yljga_record_count.csv (Task 2)")
print("- yljga_metrics.csv (Task 3)")
print("- yljga_type_metrics.csv (Task 4)")
```
