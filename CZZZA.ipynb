```python
import pandas as pd

# 读取CSV文件
try:
    df = pd.read_csv('E:\CZZZA2.csv')
    print("文件已成功读取。")
except FileNotFoundError:
    print("错误：文件 'CZZZA.csv' 未找到。请确保文件与脚本在同一目录下。")
    exit()

# 1、除去表头的总记录数
total_records = len(df)
print(f"1、除去表头的总记录数: {total_records}\n")

# 2、根据椭球面积字段，求和
total_area = df['椭球面积'].sum()
print(f"2、椭球面积字段的总和: {total_area}\n")

# 按'XZQ'字段进行分组
grouped_by_xzq = df.groupby('XZQ')

print("=== 按XZQ进行分组统计 ===")

# 遍历每个分组
for xzq, group in grouped_by_xzq:
    print(f"\n--- XZQ: {xzq} ---")

    # 3、统计CC字段中，值为070101的记录数
    records_070101 = group[group['CC'] == 70101].shape[0]
    print(f"3、CC字段中，值为070101的记录数: {records_070101}")

    # 4、统计CC字段中，值为070101的所有记录椭球面积字段的总和
    total_area_070101 = group[group['CC'] == 70101]['椭球面积'].sum()
    print(f"4、CC字段中，值为070101的所有记录椭球面积字段的总和: {total_area_070101}")

    # 5、统计CC字段中，值为070102的记录数
    records_070102 = group[group['CC'] == 70102].shape[0]
    print(f"5、CC字段中，值为070102的记录数: {records_070102}")

    # 6、统计CC字段中，值为070102的所有记录椭球面积字段的总和
    total_area_070102 = group[group['CC'] == 70102]['椭球面积'].sum()
    print(f"6、CC字段中，值为070102的所有记录椭球面积字段的总和: {total_area_070102}")
```
