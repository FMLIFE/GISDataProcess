```python
import pandas as pd

# Load the datasets
xxa_df = pd.read_csv('E:\csv-utf8\XXA.csv')
xxp_df = pd.read_csv('E:\csv-utf8\XXP.csv')

# --- 1. Total records grouped by XZQ and CC from both files ---
combined_df = pd.concat([xxa_df, xxp_df], ignore_index=True)
total_records_by_xzq_cc = combined_df.groupby(['XZQ', 'CCN']).size().reset_index(name='Total_Records')
total_records_by_xzq_cc.to_csv(r"E:\csv-utf8\结果\total_records_by_xzq_cc.csv", index=False, encoding='utf-8-sig')
print("1. Total records from both files, grouped by 'XZQ' and 'CC' saved to total_records_by_xzq_cc.csv")

# --- 2. XXA records grouped by XZQ and CC, with total count ---
xxa_records_by_xzq_cc = xxa_df.groupby(['XZQ', 'CCN']).size().reset_index(name='Total_Records_XXA')
xxa_records_by_xzq_cc.to_csv(r"E:\csv-utf8\结果\xxa_records_by_xzq_cc.csv", index=False, encoding='utf-8-sig')
print("2. Total records from XXA file, grouped by 'XZQ' and 'CC' saved to xxa_records_by_xzq_cc.csv")

# --- 3. XXA records grouped by XZQ and CC, with sum of '椭球面积' and record count ---
xxa_summary_by_xzq_cc = xxa_df.groupby(['XZQ', 'CCN']).agg(
    total_area=('椭球面积', 'sum'),
    record_count=('CC', 'size')
).reset_index()
xxa_summary_by_xzq_cc.to_csv(r"E:\csv-utf8\结果\xxa_summary_by_xzq_cc.csv", index=False, encoding='utf-8-sig')
print("3. XXA records grouped by 'XZQ' and 'CC', with total '椭球面积' and record count saved to xxa_summary_by_xzq_cc.csv")

# --- 4. XXA records grouped by XZQ, CC, then by TYPE, with sum of '椭球面积' ---
xxa_summary_by_xzq_cc_type = xxa_df.groupby(['XZQ', 'CCN', 'TYPE']).agg(
    total_area=('椭球面积', 'sum')
).reset_index()
xxa_summary_by_xzq_cc_type.to_csv(r"E:\csv-utf8\结果\xxa_summary_by_xzq_cc_type.csv", index=False, encoding='utf-8-sig')
print("4. XXA records grouped by 'XZQ', 'CC' and 'TYPE', with total '椭球面积' saved to xxa_summary_by_xzq_cc_type.csv")

print("\nAll statistical results have been successfully written to separate CSV files.")
```
