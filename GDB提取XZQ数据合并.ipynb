```python
import arcpy
import os

# ================= 配置区域 =================
# 存放这23个GDB的父文件夹路径
input_folder = r"D:\Pro Projects\2026不动产数据处理\第一批" 

# 输出结果的GDB路径（需提前建好或指定一个现有的GDB）
output_gdb = r"D:\Pro Projects\2026不动产数据处理\XZQCD.gdb"

# 合并后的图层名称
output_fc_name = "XZQCD"
# ============================================

def merge_xzq_layers():
    output_path = os.path.join(output_gdb, output_fc_name)
    
    # 将工作空间设置为包含GDB的目录
    arcpy.env.workspace = input_folder
    
    # 查找目录下的所有 FileGDB
    gdb_list = arcpy.ListWorkspaces("*", "FileGDB")
    
    if not gdb_list:
        print(f"错误: 在 {input_folder} 中未找到任何 GDB 文件。")
        return

    xzq_layers_to_merge = []
    
    # 遍历每个GDB，构建XZQ图层的绝对路径
    for gdb in gdb_list:
        xzq_path = os.path.join(gdb, "XZQ")
        
        # 检查 XZQ 图层在当前 GDB 中是否存在
        if arcpy.Exists(xzq_path):
            xzq_layers_to_merge.append(xzq_path)
        else:
            print(f"警告: {gdb} 中未找到 XZQ 图层，已跳过。")
            
    if not xzq_layers_to_merge:
        print("错误: 没有找到任何有效的 XZQ 图层进行合并。")
        return
        
    print(f"共找到 {len(xzq_layers_to_merge)} 个 XZQ 图层，开始执行合并...")
    
    try:
        # 执行合并操作
        arcpy.management.Merge(xzq_layers_to_merge, output_path)
        print(f"✅ 合并成功！结果已输出至: {output_path}")
    except arcpy.ExecuteError:
        print("❌ 合并过程中发生错误:")
        print(arcpy.GetMessages(2))
    except Exception as e:
        print(f"❌ 发生未知错误: {e}")

if __name__ == "__main__":
    merge_xzq_layers()
```
