```python
import os
import shutil
import argparse
from pathlib import Path

def move_dwg_files(source_dir, dest_base):
    # 获取所有dwg文件
    dwg_files = [f for f in os.listdir(source_dir) if f.lower().endswith('.dwg')]
    
    if not dwg_files:
        print("未找到DWG文件！")
        return
    
    # 创建基础目标目录（如果不存在）
    os.makedirs(dest_base, exist_ok=True)
    
    # 计算需要创建的文件夹数量
    folder_count = (len(dwg_files) + 499) // 500
    
    # 处理每个文件夹
    for folder_num in range(1, folder_count + 1):
        # 创建文件夹
        folder_name = f"{folder_num:03d}"  # 格式化为三位数，如001, 002等
        folder_path = os.path.join(dest_base, folder_name)
        os.makedirs(folder_path, exist_ok=True)
        
        # 获取当前文件夹需要处理的文件
        start_idx = (folder_num - 1) * 500
        end_idx = min(start_idx + 500, len(dwg_files))
        current_files = dwg_files[start_idx:end_idx]
        
        # 移动文件
        for file in current_files:
            source_path = os.path.join(source_dir, file)
            dest_path = os.path.join(folder_path, file)
            try:
                shutil.move(source_path, dest_path)
                print(f"已移动: {file} -> {folder_name}")
            except Exception as e:
                print(f"移动失败 {file}: {e}")
    
    print(f"处理完成！共创建了 {folder_count} 个文件夹。")

def main():
    parser = argparse.ArgumentParser(description='将DWG文件按每500个一组移动到不同的文件夹')
    parser.add_argument('--source', required=True, help='源文件夹路径')
    parser.add_argument('--dest', required=True, help='目标基础文件夹路径')
    args = parser.parse_args()
    
    # 确保路径正确
    source_dir = os.path.abspath(args.source)
    dest_base = os.path.abspath(args.dest)
    
    if not os.path.exists(source_dir):
        print(f"源文件夹不存在: {source_dir}")
        return
    
    move_dwg_files(source_dir, dest_base)

if __name__ == "__main__":
    main()    
```
