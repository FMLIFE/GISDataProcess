```python
import arcpy
import os

def feature_classes_to_excel(gdb_path, output_folder):
    """
    批量将FileGDB中的所有要素类转换为Excel文件（.xlsx）。

    Args:
        gdb_path (str): FileGDB的完整路径。
        output_folder (str): 用于保存Excel文件的输出文件夹路径。
    """
    try:
        # 设置工作空间
        arcpy.env.workspace = gdb_path
        
        # 确保输出文件夹存在
        if not os.path.exists(output_folder):
            os.makedirs(output_folder)
            print(f"已创建输出文件夹: {output_folder}")

        print(f"正在处理GDB: {gdb_path}\n")
        
        # 遍历GDB中的所有要素数据集
        datasets = arcpy.ListDatasets(feature_type='Feature')
        if not datasets:
            datasets = ['']  # 如果没有要素数据集，则处理GDB根目录

        for dataset in datasets:
            # 遍历要素数据集或GDB根目录下的所有要素类
            current_workspace = os.path.join(gdb_path, dataset) if dataset else gdb_path
            arcpy.env.workspace = current_workspace
            feature_classes = arcpy.ListFeatureClasses()

            for fc in feature_classes:
                # 构造输出Excel文件名
                # os.path.splitext() 可以处理文件扩展名，这里我们不需要
                output_name = "b_" + os.path.splitext(fc)[0] + ".xlsx"
                output_excel = os.path.join(output_folder, output_name)

                print(f"正在将要素类 '{fc}' 转换为 '{output_excel}'...")
                
                # 使用Table To Excel工具
                arcpy.conversion.TableToExcel(
                    Input_Table=fc,
                    Output_Excel_File=output_excel,
                    Use_field_alias_as_column_header="NAME",
                    Use_domain_and_subtype_description="CODE"
                )
                
                print(f"  转换成功。\n")

        print("所有要素类已成功转换为Excel文件。")

    except arcpy.ExecuteError:
        print(f"ArcPy执行错误: {arcpy.GetMessages(2)}")
    except Exception as e:
        print(f"发生了一个错误: {e}")

if __name__ == '__main__':
    # 请修改为你的FileGDB完整路径
    gdb_path = r"D:\MyProject\MyProject2.gdb" 
    
    # 请修改为你的输出文件夹路径
    output_folder = r"D:\MyProject\excel0907" 
    
    feature_classes_to_excel(gdb_path, output_folder)
```
