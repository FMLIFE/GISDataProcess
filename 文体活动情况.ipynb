```python
import pandas as pd
import io

# Load the two CSV files into pandas DataFrames
try:
    df_a = pd.read_csv(r"E:\csv-utf8\WHHDA.csv")
    df_p = pd.read_csv(r"E:\csv-utf8\WHHDP.csv")
except Exception as e:
    print(f"Error loading files: {e}")
    # Handle the error or exit
    exit()

# Create a list to store the results of each step
results = []

# 1. Statistical analysis of both files
# Combine the two dataframes
df_combined = pd.concat([df_a, df_p])
# Group by XZQ and count the total number of records
grouped_combined = df_combined.groupby('XZQ').size().reset_index(name='Total_Count')
# Add a column to identify the task
grouped_combined['Task'] = 'Task 1: Combined Record Count'
results.append(grouped_combined)

# 2. Statistical analysis of WHHDA file
# Group by XZQ, calculate the sum of '椭球面积' and the total number of records
grouped_a_metrics = df_a.groupby('XZQ').agg(
    椭球面积_sum=('椭球面积', 'sum'),
    Record_Count=('OBJECTID', 'count')
).reset_index()
grouped_a_metrics['Task'] = 'Task 2: WHHDA Metrics by XZQ'
results.append(grouped_a_metrics)

# Combine all results into a single DataFrame
final_output = pd.concat(results, ignore_index=True)

# Save the combined DataFrame to a single CSV file without a header for the intermediate sections
final_output.to_csv(r"E:\csv-utf8\结果\文化艺术场馆.csv", index=False, encoding='utf-8-sig')

print("All statistical results have been saved to 'statistical_results.csv'.")
print("The file contains the results of all three tasks combined.")

# Load the two CSV files into pandas DataFrames
try:
    df_a = pd.read_csv(r"E:\csv-utf8\TYHDA.csv")
    df_p = pd.read_csv(r"E:\csv-utf8\TYHDP.csv")
except Exception as e:
    print(f"Error loading files: {e}")
    # Handle the error or exit
    exit()

# Create a list to store the results of each step
results = []

# 1. Statistical analysis of both files
# Combine the two dataframes
df_combined = pd.concat([df_a, df_p])
# Group by XZQ and count the total number of records
grouped_combined = df_combined.groupby('XZQ').size().reset_index(name='Total_Count')
# Add a column to identify the task
grouped_combined['Task'] = 'Task 1: Combined Record Count'
results.append(grouped_combined)

# 2. Statistical analysis of WHHDA file
# Group by XZQ, calculate the sum of '椭球面积' and the total number of records
grouped_a_metrics = df_a.groupby('XZQ').agg(
    椭球面积_sum=('椭球面积', 'sum'),
    Record_Count=('OBJECTID', 'count')
).reset_index()
grouped_a_metrics['Task'] = 'Task 2: WHHDA Metrics by XZQ'
results.append(grouped_a_metrics)

# Combine all results into a single DataFrame
final_output = pd.concat(results, ignore_index=True)

# Save the combined DataFrame to a single CSV file without a header for the intermediate sections
final_output.to_csv(r"E:\csv-utf8\结果\体育场馆（含独立足球场）.csv", index=False, encoding='utf-8-sig')

print("All statistical results have been saved to 'statistical_results.csv'.")
print("The file contains the results of all three tasks combined.")
```
